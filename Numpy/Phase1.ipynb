# Numpy - (Numerical Python)

```python
import numpy as np
```

## Creating Arrays from List

```python
arr_1d = np.array([1,2,3,4,5])
print("1d Arrays: ", arr_1d)

arr_2d = np.array([[1,2,3],[4,5,6]])
print("2d Arrays: ",arr_2d)
```

## List v/s Numpy array

```python
py_list = [2,3,4] #list multiplication - double the list
print("List Multiplication result:", py_list * 2)

np_arr = np.array([4,5,6]) #element wise multiplication
print("Numpy Array Multiplication result:", np_arr * 2)
```

### Efficency of List and Numpy array

```python
import time
start = time.time()
py_list = [i*2 for i in range(1000000)]
print("Python List Time:", time.time() - start)

start1 = time.time()
np_arr = np.arange(1000000) * 2
print("Numpy Array Time:", time.time() - start1)
```

## Creating array from scratch

```python
zeros = np.zeros((3,4))
print("Zeros array:\n", zeros)

ones = np.ones((2,3))
print("Ones array:\n", ones)

full = np.full((2,2),1) # one dimentional array of 2 elements with all values as 1
print("Full array:\n", full)

random = np.random.random((2,2)) #2x2 array with random values
print("Random array:\n", random)

sequence = np.arange(10,20,2) #start, end, step
print("Sequence array:\n", sequence)
```

## Vector, Matrix and Tensor

```python
vector = np.array([1,2,3,4,5])
print("Array:", vector)
print("Shape of array:", vector.shape)

matrix = np.array([[1,2,3],[4,5,6]])
print("Matrix:\n", matrix)
print("Shape of matrix:", matrix.shape)

tensor = np.array([[1,2],[3,4],[5,6]])
print("Tensor:\n", tensor)
print("Shape of tensor:", tensor.shape)
```

### Array Properties

```python
new_arr = np.array([[1,2,3], [3,4,5]])
print("Shape:", new_arr.shape)
print("Dimensions:", new_arr.ndim)
print("Data Type:", new_arr.dtype)
print("Size:", new_arr.size)
```

## Array Reshaping

```python
arr = np.arange(10)
print("Original array:", arr)

reshaped = arr.reshape(2,5)
print("Reshaped array (2x5:\n", reshaped)

flattened = reshaped.flatten()
print("Flattened array:", flattened)

raveled = reshaped.ravel() #returns view instead of copy
print("Raveled array:", raveled)

transpose = reshaped.T #transpose: rows become columns and vice versa
print("Transposed array:\n", transpose)
```
